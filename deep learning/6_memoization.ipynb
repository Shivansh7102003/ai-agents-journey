```python
import time
def fib(n):
    if n==0 or n==1:
        return 1
    else:
        return fib(n-1) + fib(n-2)
    
```

```python
# exponential time complexity
start = time.time()

fib(40)

end = time.time()

total_time = end - start
print(total_time)
```

# memoization

```python
def fib1(n , d):
    if n in d :
        return d[n]
    else:
        d[n] = fib1(n-1,d) + fib1(n-2,d)
        return d[n]
    
```

```python
start1 = time.time()
d = {0:1 , 1:1}
fib1(1000, d)
print(time.time()-start1)
# print(d)
```
